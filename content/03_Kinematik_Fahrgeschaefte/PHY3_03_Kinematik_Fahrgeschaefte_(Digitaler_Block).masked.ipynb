## 1. Räumliche Kinematik "Top-Scan"

## 1.1 System

Das Fahrgeschäft Top-Scan besteht aus einem feststehenden aber geneigten Mast, einem rotierenden Arm an dessen Ende ein rotierneder Stern befestigt ist an welchem wiederum sechs rotierende Sitzbänke befestigt sind. Alle Körper werden starr modelliert und jedem Körper ist ein körperfestes Koordinatensystem zugeordnet. Konkret ist dem Mast das Systems $M (\boldsymbol{e}_{x}^M, \boldsymbol{e}_{y}^M, \boldsymbol{e}_{z}^M)$, dem Arm das System $A (\boldsymbol{e}_{x}^A, \boldsymbol{e}_{y}^A, \boldsymbol{e}_{z}^A)$, dem Stern das System $S (\boldsymbol{e}_{x}^S, \boldsymbol{e}_{y}^S, \boldsymbol{e}_{z}^S)$ und einer Sitzbank das System $B (\boldsymbol{e}_{x}^B, \boldsymbol{e}_{y}^B, \boldsymbol{e}_{z}^B)$ zugeordnet. Der Mast ist fest mit dem inertialen Boden verbunden aber um den Winkel $\alpha = 45°$ um die $\boldsymbol{e}_{x}^I$ Achse geneigt. Im Punkt $M$ ist der Arm der Länge $L$ befestigt, der mit $\beta(t)$ um die $\boldsymbol{e}_{z}^M$ Achse rotiert. Die Drehachse des Sterns ist ebenfalls um 45° gegenüber der $\boldsymbol{e}_{x}^A$ Achse gedreht. Diese "statische" Verdrehung wird über das Hilfssystem $H (\boldsymbol{e}_{x}^H, \boldsymbol{e}_{y}^H, \boldsymbol{e}_{z}^H)$ beschrieben, welches gegenüber $A$ um die gemeinsame $\boldsymbol{e}_{x}$ Achse gedreht ist. Der Stern rotiert mit $\delta(t)$ um die mit $H$ gemeinsame $\boldsymbol{e}_{z}$ Achse. 
Die Sitzbank ist in radialer Richtung vom Stern, also $\boldsymbol{e}_{x}$ Richtung, ausgerichtet aber um den Winkel $\varphi(t)$ um diese Achse verdreht. Im Punkt $P$ sitzt eine Person auf der Sitzbank deren Kinemaitk im System $B$ beschrieben wird.
